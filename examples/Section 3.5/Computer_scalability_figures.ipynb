```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ============================================================
# Data
# ============================================================

grid_labels = [
    r"$12\times12\times6$",
    r"$24\times24\times12$",
    r"$48\times48\times24$"
]

n_cells = np.array([
    12 * 12 * 6,
    24 * 24 * 12,
    48 * 48 * 24
])

memory = np.array([
    0.37,
    1.75,
    47.86
])

time_min = np.array([
    94 / 60,
    18824 / 60,
    242271 / 60
])


# ============================================================
# Colors
# ============================================================

blue = "#0072B2"
gray = "#555555"


# ============================================================
# Figure
# ============================================================

fig = plt.figure(figsize=(11, 4.8))
plt.subplots_adjust(wspace=0.3)


# ============================================================
# (a) Maximum memory
# ============================================================

ax = fig.add_subplot(121)

# Measured data
ax.loglog(
    n_cells,
    memory,
    color=blue,
    marker="o",
    markersize=7,
    markerfacecolor="white",
    markeredgecolor=blue,
    markeredgewidth=1.8,
    linewidth=1.8,
    zorder=3
)


# ------------------------------------------------------------
# Grid-size annotations
# ------------------------------------------------------------

memory_offsets = [
    (8, 8),
    (8, 8),
    (-72, -22)
]


# Labels
ax.set_xlabel(
    "Number of model cells",
    fontsize=12
)

ax.set_ylabel(
    "Maximum memory (GB)",
    fontsize=12
)

# Panel label
#ax.text(
#    0.03, 0.97,
#    "a",
#    transform=ax.transAxes,
#    fontsize=13,
#    fontweight="bold",
#    va="top"
#)

# Major grid
ax.grid(
    True,
    which="major",
    linestyle="--",
    linewidth=0.65,
    color="#B0B0B0",
    alpha=0.6
)

# Minor grid
ax.grid(
    True,
    which="minor",
    linestyle=":",
    linewidth=0.45,
    color="#C8C8C8",
    alpha=0.5
)

# Ticks
ax.tick_params(
    axis="both",
    which="major",
    labelsize=10.5,
    #direction="in",
    length=5,
    width=0.9,
    top=True,
    right=True
)

ax.tick_params(
    axis="both",
    which="minor",
    #direction="in",
    length=3,
    width=0.7,
    top=True,
    right=True
)

ax.set_xticks(n_cells)
ax.set_xticklabels(
    grid_labels,
    fontsize=10
)

# Spine
for spine in ax.spines.values():
    spine.set_linewidth(0.9)


# ============================================================
# (b) Total computational time
# ============================================================

ax = fig.add_subplot(122)

ax.loglog(
    n_cells,
    time_min,
    color=blue,
    marker="o",
    markersize=7,
    markerfacecolor="white",
    markeredgecolor=blue,
    markeredgewidth=1.8,
    linewidth=1.8,
    zorder=3
)

# Labels
ax.set_xlabel(
    "Number of model cells",
    fontsize=12
)

ax.set_ylabel(
    "Total computational time (min)",
    fontsize=12
)

# Panel label
#ax.text(
#    0.03, 0.97,
#    "b",
#    transform=ax.transAxes,
#    fontsize=13,
#    fontweight="bold",
#    va="top"
#)

# Major grid
ax.grid(
    True,
    which="major",
    linestyle="--",
    linewidth=0.65,
    color="#B0B0B0",
    alpha=0.6
)

# Minor grid
ax.grid(
    True,
    which="minor",
    linestyle=":",
    linewidth=0.45,
    color="#C8C8C8",
    alpha=0.5
)

# Ticks
ax.tick_params(
    axis="both",
    which="major",
    labelsize=10.5,
    #direction="in",
    length=5,
    width=0.9,
    top=True,
    right=True
)

ax.tick_params(
    axis="both",
    which="minor",
    #direction="in",
    length=3,
    width=0.7,
    top=True,
    right=True
)

ax.set_xticks(n_cells)
ax.set_xticklabels(
    grid_labels,
    fontsize=10
)

# Spine
for spine in ax.spines.values():
    spine.set_linewidth(0.9)

plt.savefig("1011_computational_scalability.svg", dpi = 600, bbox_inches = "tight")
```
