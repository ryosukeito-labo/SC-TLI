### SC-TLI
L1-L2 penalty in the space domain and structural coupling in the temporal domain

$$
P(\boldsymbol{\beta}) = 
\lambda_s \left( \alpha \left\| \boldsymbol{\beta} \right\|_1 + \frac{1 - \alpha}{2} \| \boldsymbol{\beta} \|_2^2 \right)
+
\lambda_t \sum_{\mathcal{G}_j} w_{\mathcal{G}_j} \left\| \left( \boldsymbol{D} \boldsymbol{\beta} \right)_{\mathcal{G}_j} \right\|_2
$$

```python
# data misfit
def misfit(f, K, beta):
    return 0.5 * np.linalg.norm(f - K.dot(beta))**2

# penalty
def pen1(beta, alpha):
    return alpha * np.linalg.norm(beta, ord = 1) + (1 - alpha) * 0.5 * np.linalg.norm(beta)**2
    
def pen2(Dbeta, sw, num_of_cells, times):
    
    pen2 = 0
    
    for k in range(num_of_cells):
        temp = 0
        
        for j in range(times-1):
            temp += Dbeta[k + j*num_of_cells]**2
            
        pen2 += sw[k] * np.sqrt(temp)
        
    return pen2
```

```python
def calcCurvature2(x,y):
    dif_x = np.gradient(x)
    dif_y = np.gradient(y)
    dif2_x = np.gradient(dif_x)
    dif2_y = np.gradient(dif_y)
    curvature = np.abs(dif2_x * dif_y - dif_x * dif2_y) / (dif_x **2 + dif_y **2)**1.5
    return curvature
```

### Kernel matrix

```python
from numba import njit, prange

@njit(parallel=True, fastmath=True)
def kernel_matrix2(
    exf,
    mgz,
    xobs,
    yobs,
    zobs,
    xgrd,
    ygrd,
    zgrd,
    dim,
    L
):

    nobs  = xobs.shape[0]
    ncell = xgrd.shape[0]

    xmin = np.min(xgrd)
    xmax = np.max(xgrd)

    ymin = np.min(ygrd)
    ymax = np.max(ygrd)

    K = np.zeros((nobs, ncell))

    for s in prange(ncell):

        dx1 = dim[0,0]
        dx2 = dim[0,1]

        dy1 = dim[1,0]
        dy2 = dim[1,1]

        dz1 = dim[2,0]
        dz2 = dim[2,1]

        if abs(xgrd[s] - xmin) < 1e-10:
            dx1 += L
        elif abs(xgrd[s] - xmax) < 1e-10:
            dx2 += L

        if abs(ygrd[s] - ymin) < 1e-10:
            dy1 += L
        elif abs(ygrd[s] - ymax) < 1e-10:
            dy2 += L

        dim_new = np.empty((3,2))

        dim_new[0,0] = dx1
        dim_new[0,1] = dx2

        dim_new[1,0] = dy1
        dim_new[1,1] = dy2

        dim_new[2,0] = dz1
        dim_new[2,1] = dz2

        h = magcal_numba.prism(
            mgz,
            xobs,
            yobs,
            zobs,
            xgrd[s],
            ygrd[s],
            zgrd[s],
            dim_new
        )

        K[:,s] = magcal_numba.total_force(exf, h)

        if s%1000 == 0: 
            print(s) 

    return K
```

### Input models

```python
### Magnetization structure  ###

# ------------------------------------------------------------ #

# Time 1
beta1 = 0 * np.ones(len(xgrd))
for iz in range(nz):
    for iy in range(ny):
        for ix in range(nx):
            k = ix + iy*nx + iz*nx*ny

            # 東: migration
            if (0.2 < xgrd[k]) & (xgrd[k] < 0.8) & (-0.3 < ygrd[k]) & (ygrd[k] < 0.3) & (-1.0 < zgrd[k]) & (zgrd[k] < -0.4):
                beta1[k] = -4

            # 西: diffusion
            if (-1.4 < xgrd[k]) & (xgrd[k] < -0.6) & (-0.4 < ygrd[k]) & (ygrd[k] < 0.4) & (-1.7 < zgrd[k]) & (zgrd[k] < -0.9):
                beta1[k] = -3

# Time 2
beta2 = 0 * np.ones(len(xgrd))
for iz in range(nz):
    for iy in range(ny):
        for ix in range(nx):
            k = ix + iy*nx + iz*nx*ny

            # 東: migration
            if (0.4 < xgrd[k]) & (xgrd[k] < 1.0) & (-0.3 < ygrd[k]) & (ygrd[k] < 0.3) & (-1.0 < zgrd[k]) & (zgrd[k] < -0.4):
                beta2[k] = -4

            # 西: diffusion
            if (-1.4 < xgrd[k]) & (xgrd[k] < -0.6) & (-0.4 < ygrd[k]) & (ygrd[k] < 0.4) & (-1.7 < zgrd[k]) & (zgrd[k] < -0.7):
                beta2[k] = -3


# Time 3
beta3 = 0 * np.ones(len(xgrd))
for iz in range(nz):
    for iy in range(ny):
        for ix in range(nx):
            k = ix + iy*nx + iz*nx*ny

            # 東: migration
            if (0.6 < xgrd[k]) & (xgrd[k] < 1.2) & (-0.3 < ygrd[k]) & (ygrd[k] < 0.3) & (-1.0 < zgrd[k]) & (zgrd[k] < -0.4):
                beta3[k] = -4

            # 西: diffusion
            if (-1.4 < xgrd[k]) & (xgrd[k] < -0.6) & (-0.4 < ygrd[k]) & (ygrd[k] < 0.4) & (-1.7 < zgrd[k]) & (zgrd[k] < -0.5):
                beta3[k] = -3
                
# Time 4
beta4 = 0 * np.ones(len(xgrd))
for iz in range(nz):
    for iy in range(ny):
        for ix in range(nx):
            k = ix + iy*nx + iz*nx*ny

            # 東: migration
            if (0.8 < xgrd[k]) & (xgrd[k] < 1.4) & (-0.3 < ygrd[k]) & (ygrd[k] < 0.3) & (-1.0 < zgrd[k]) & (zgrd[k] < -0.4):
                beta4[k] = -4

            # 西: diffusion
            if (-1.4 < xgrd[k]) & (xgrd[k] < -0.6) & (-0.4 < ygrd[k]) & (ygrd[k] < 0.4) & (-1.7 < zgrd[k]) & (zgrd[k] < -0.3):
                beta4[k] = -3
                
bb1 = beta1.reshape(nz, ny, nx)
bb2 = beta2.reshape(nz, ny, nx)
bb3 = beta3.reshape(nz, ny, nx)
bb4 = beta4.reshape(nz, ny, nx)

# ------------------------------------------------------------ #

# Magnetic anomaly calculation
f1_row = K1.dot(beta1)
f2_row = K1.dot(beta2)
f3_row = K1.dot(beta3)
f4_row = K1.dot(beta4)

# Centering
f1_row = f1_row - f1_row.mean()
f2_row = f2_row - f2_row.mean()
f3_row = f3_row - f3_row.mean()
f4_row = f4_row - f4_row.mean()

# Add Gaussian noise
np.random.seed(42)
noise1 = np.random.normal(loc=0, scale=5.0, size=len(f1_row))  # 5 nT
noise2 = np.random.normal(loc=0, scale=5.0, size=len(f2_row))  # 5 nT
noise3 = np.random.normal(loc=0, scale=5.0, size=len(f3_row))  # 5 nT
noise4 = np.random.normal(loc=0, scale=5.0, size=len(f4_row))  # 5 nT

fobs1 = f1_row + noise1
fobs2 = f2_row + noise2
fobs3 = f3_row + noise3
fobs4 = f4_row + noise4
```

```python
# hyper-parameters
lams_t = np.power(10., np.arange(0.0, 2.1, 0.25))
print("lams_t = {}".format(lams_t))

lam_s = 31.6227766
alpha = 0.9
```

```python
%%time
params_all = []
betas1_all = []
betas2_all = []
betas3_all = []
betas4_all = []

for lit in range(len(lams_t)):

    params_all.append([lam_s, lams_t[lit]])
    print("lm1 = {:.3f}, lm2 = {:.3f}".format(lam_s, lams_t[lit]))

    model = ADMMGroupLasso(lambda_s = lam_s, 
                           lambda_t = lams_t[lit], 
                           alpha = alpha, 
                           rho = rho, 
                           eta = eta, 
                           tol = 1e-3
                          )
    model.fit(fobs, 
              Z, 
              D, 
              sw, 
              inv1, 
              inv2, 
              times = 4, 
              num_of_cells = nx*ny*nz
             )
    
    beta_sim1 = model.beta_[0*nx*ny*nz:1*nx*ny*nz] / w1
    beta_sim2 = model.beta_[1*nx*ny*nz:2*nx*ny*nz] / w1
    beta_sim3 = model.beta_[2*nx*ny*nz:3*nx*ny*nz] / w1
    beta_sim4 = model.beta_[3*nx*ny*nz:4*nx*ny*nz] / w1

    betas1_all.append(beta_sim1)
    betas2_all.append(beta_sim2)
    betas3_all.append(beta_sim3)
    betas4_all.append(beta_sim4)

# ndarry
params_all = np.array(params_all)
betas1_all = np.array(betas1_all)
betas2_all = np.array(betas2_all)
betas3_all = np.array(betas3_all)
betas4_all = np.array(betas4_all)
```
